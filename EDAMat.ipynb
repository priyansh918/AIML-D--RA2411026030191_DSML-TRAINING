```python
#IMPORT REQUIRED LIBRARIES
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#load the dataset
df=pd.read_csv("/content/salary_dataset (1) (1).csv")
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.describe()
```

```python
#checking for missing values

df.isnull().sum()
```

```python
#Univariate analysis
#Distribution of employess by salary

plt.figure(figsize=(8,5))
sns.histplot(df['Salary'], bins=10, kde=True)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Numbe of Employees")
plt.show()
```

```python
#Experience Distribution

plt.figure(figsize=(8,5))
sns.histplot(df['Experience'], bins=10, kde=True)
plt.title("Experience Distribution")
plt.xlabel("Years of Experience")
plt.ylabel("Number of Employees")
plt.show()
```

```python
#Number of employees by jobrole
plt.figure(figsize=(10,5))

sns.countplot(data=df, x="JobRole")

plt.title("Average salary by Job Role")
plt.xlabel("Job Role")
plt.ylabel("number of Employees")
plt.xticks(rotation=45)
plt.show()
```

```python
#Average Salary by JobRole
#Bi-variate analysis(including two variables)

plt.figure(figsize=(12,8))

sns.barplot(data=df, x="JobRole", y="Salary")

plt.title("Average salary by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Average salary")
plt.xticks(rotation=45)
plt.show()
```

```python
#Salary by Education
plt.figure(figsize=(8,5))

sns.barplot(data=df, x="Education", y="Salary")

plt.title("Average salary by Education")
plt.xlabel("Educat")
plt.ylabel("Average salary")
plt.xticks(rotation=45)
plt.show()
```
